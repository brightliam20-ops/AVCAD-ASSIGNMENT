### Extracting Text from CV PDF

Now, let's extract the text content from your provided CV PDF file.

```python
import PyPDF2

# Path to the uploaded CV file
cv_file_path = '/content/Olaniyan_Samuel_CV.pdf'

# Extract text from PDF
with open(cv_file_path, 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    text = ''
    for page_num in range(len(reader.pages)):
        page = reader.pages[page_num]
        text += page.extract_text() if page.extract_text() else ''

cv_text = text

if not cv_text:
    print("Could not extract text from the PDF. Please ensure the PDF is not image-based or protected.")
else:
    print("Text extracted successfully. First 500 characters:\n", cv_text[:500])
```

### Generating the Word Cloud

Now, let's generate and display the word cloud from your CV text. We'll perform some basic text cleaning by converting the text to lowercase and removing common English stop words.

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Initialize stopwords with default English stopwords
stopwords = set(STOPWORDS)
# The previous custom stopwords list was too aggressive, removing many relevant CV words.
# stopwords.update(["experience", "skills", "project", "management", "developed", "responsibilities", "used", "work", "team", "data", "python", "sql", "tools", "etc", "ability", "demonstrated", "strong", "knowledge", "proficient", "effective", "related", "including", "key", "environment"])

# Generate a word cloud image
wordcloud = WordCloud(width = 800, height = 800,
                background_color ='white',
                stopwords = stopwords,
                min_font_size = 10).generate(cv_text)

# Display the generated image:
plt.figure(figsize = (8, 8), facecolor = None)
plt.imshow(wordcloud)
plt.axis("off")
plt.tight_layout(pad = 0)

plt.show()
```
